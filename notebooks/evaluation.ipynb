```python
import mlflow
import pandas as pd
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix
```

### Importing the 'Champion' mmodel from registry

```python
mlflow.set_tracking_uri('sqlite:///../mlflow.db')
champion_model = mlflow.sklearn.load_model("models:/Champion/latest")
```

### Evaluating on train data for testing

```python
X_processed_csv_path = '../data/pre_processed/train.csv'
X_raw_csv_path = '../data/raw/train.csv'
X_processed = pd.read_csv(X_processed_csv_path)
y_orig = pd.read_csv(X_raw_csv_path).Transported
```

```python
y_prob = champion_model.predict_proba(X_processed)
current_threshold = 0.5584083119472486
y_pred = y_prob[:,1] > current_threshold
```

```python
print(f"the accureacy is: {accuracy_score(y_orig,y_pred)}")
print(f"the roc auc score is: {roc_auc_score(y_orig,y_prob[:,1])}")
```
